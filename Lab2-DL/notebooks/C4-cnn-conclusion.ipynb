## 🔬 Advanced CNN Architectures in Medicine

### 🏗️ ResNet (Residual Networks)

**Why important in medicine:**
- **Deep networks**: Can learn complex patterns
- **Gradient flow**: Avoids vanishing gradient problem
- **Transfer learning**: Use pretrained models

**Medical applications:**
- **Radiology**: Chest X-ray analysis
- **Pathology**: Histological image analysis
- **Dermatology**: Skin lesion classification

### 🔄 U-Net (Segmentation)

**Why important in medicine:**
- **Pixel/voxel-wise classification**: Segment anatomical structures
- **Skip connections**: Preserves details
- **End-to-end training**: Optimizes for segmentation

**Medical applications:**
- **Tumor segmentation**: MRI, CT
- **Organ segmentation**: Heart, liver, kidneys
- **Vessel segmentation**: Angiography

### 🎯 Attention and Explainability Mechanisms

**Why important in medicine:**
- **Attention**: The model focuses on relevant areas
- **Explainability**: Can show where the model "looks"
- **Robustness**: Less affected by irrelevant details

**Medical applications:**
- **Lesion detection**: Focus on pathological areas
- **Multi-modal fusion**: Combines different image types
- **Temporal analysis**: Looks at changes over time

### ✅ Check Your Understanding: CNN Series Review

**Q1:** What are the three main types of layers in a CNN, and what does each do?

<details>
<summary>Click for answer</summary>

| Layer Type | Purpose |
|------------|---------|
| **Convolutional** | Extract features (edges → textures → objects) via learnable filters |
| **Pooling** | Reduce spatial dimensions, add translation invariance |
| **Fully Connected** | Combine features for final classification |

Together, these form a **feature hierarchy**: early layers detect simple patterns, deep layers detect complex concepts.
</details>

**Q2:** Why is explainability (e.g., Grad-CAM) especially important for medical AI compared to other applications?

<details>
<summary>Click for answer</summary>

**Three key reasons:**

1. **Clinical trust:** Doctors need to understand WHY the AI made a recommendation before acting on it
2. **Error detection:** Can identify if the model learned spurious correlations (e.g., hospital logo instead of pathology)
3. **Regulatory requirement:** FDA/CE approval often requires demonstrating the model focuses on clinically relevant features

In low-stakes applications (e.g., flower classification), a wrong prediction is merely inconvenient. In medicine, it could be life-threatening.
</details>

**Q3:** When would you use U-Net instead of a classification CNN like ResNet?

<details>
<summary>Click for answer</summary>

| Use Classification CNN (ResNet) | Use U-Net |
|--------------------------------|-----------|
| Need a **single label** per image | Need to identify **exact location** of structures |
| "Is this a tumor?" (yes/no) | "Where exactly is the tumor?" (pixel mask) |
| "What flower is this?" | "Outline the heart chambers" |

**U-Net is for segmentation**: when you need to know not just "what" but "where" at pixel-level precision.

Medical examples: tumor boundaries, organ segmentation, vessel tracing.
</details>

**Q4:** What is the difference between error consequences in flower classification vs. medical diagnosis?

<details>
<summary>Click for answer</summary>

| Aspect | Flower Classification | Medical Diagnosis |
|--------|----------------------|-------------------|
| **False positive** | Minor inconvenience | Unnecessary treatment, anxiety |
| **False negative** | Missed aesthetic opportunity | Missed disease, delayed treatment |
| **Stakes** | Aesthetic/hobby | Life and death |
| **Accountability** | None | Legal, ethical, professional |
| **Regulation** | None | FDA, CE marking, HIPAA |

This is why medical AI requires higher accuracy thresholds, extensive validation, and explainability.
</details>

## 📚 Glossary of Key Terms

### CNN Architecture

| Term | Definition |
|------|------------|
| **Convolution** | Operation that applies learned filters to detect features |
| **Feature map** | Output of a convolutional layer; represents detected features |
| **Pooling** | Downsampling operation (max, average) to reduce dimensions |
| **Stride** | Step size when sliding the convolution kernel |
| **Padding** | Adding zeros around input to control output size |
| **Receptive field** | Input region that influences a particular output unit |

### Advanced Architectures

| Term | Definition |
|------|------------|
| **ResNet** | Residual Network with skip connections for very deep networks |
| **U-Net** | Encoder-decoder architecture for segmentation tasks |
| **Skip connections** | Direct connections that bypass layers (help gradient flow) |
| **Transfer learning** | Using pretrained weights as starting point |
| **Attention mechanism** | Learned weights that focus on relevant regions |
| **Foundation model** | Large pretrained model adaptable to many tasks |

### Medical AI Concepts

| Term | Definition |
|------|------------|
| **Segmentation** | Pixel-wise classification (outlining structures) |
| **Classification** | Assigning a single label to an image |
| **Regression** | Predicting a continuous value (e.g., age from brain MRI) |
| **FDA clearance** | US regulatory approval for medical devices |
| **CE marking** | European regulatory approval for medical devices |
| **Clinical validation** | Testing AI in real clinical settings |

### Explainability

| Term | Definition |
|------|------------|
| **Grad-CAM** | Gradient-based visualization of important regions |
| **Saliency map** | Gradient of prediction w.r.t. input pixels |
| **Black box** | Model whose decision process is opaque |
| **Interpretable model** | Model designed to be understandable |
| **Spurious correlation** | Model learns irrelevant features that happen to correlate |

## 🏥 fastMONAI - Bergen-based Medical AI

### 🌟 What is fastMONAI?

**fastMONAI** is a Norwegian, open-source library for medical AI developed at the **Mohn Medical Imaging and Visualization Centre** / **Haukeland University Hospital**. <br>
It builds on fastai, MONAI, TorchIO and Imagedata, and simplifies the use of state-of-the-art deep learning techniques in 3D medical image analysis.

**Main goals:**
- **Simplify** the use of advanced deep learning techniques
- **Solve** classification, regression, and segmentation tasks
- **Provide** functionality for data loading, preprocessing, training, and result interpretation

### ❓ Why fastMONAI?

**Advantages:**
- **Low-code approach**: Simpler API than standard MONAI
- **Norwegian development**: Developed by Norwegian researchers (Satheshkumar Kaliyugarasan and Alexander S. Lundervold)
- **Medical focus**: Specialized for 3D medical images
- **Open source**: Free and available for everyone
- **Interactive notebooks**: Documentation as Jupyter notebooks

### 📋 Requirements

- **Python:** 3.10, 3.11, or 3.12 (Python 3.11 recommended)
- **GPU:** CUDA-compatible GPU recommended for training (CPU supported for inference)

### 🛠️ Installation

**Recommended environment setup:**
```bash
# Create conda environment
conda create -n fastmonai python=3.11
conda activate fastmonai

# Install fastMONAI
pip install fastMONAI
```

**Development installation:**
```bash
# Clone repository
git clone https://github.com/MMIV-ML/fastMONAI
cd fastMONAI

# Create development environment
conda create -n fastmonai-dev python=3.11
conda activate fastmonai-dev

# Install in development mode
pip install -e '.[dev]'
```
