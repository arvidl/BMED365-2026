## Narrow vs General AI

### 🎯 Narrow AI (Narrow AI / ANI)
- Designed for specific tasks
- All current AI is narrow AI
- Examples: Chess AI, image recognition, language translation

### 🌍 General AI (AGI)
- Can solve any intellectual task
- Human-like intelligence
- Doesn't exist yet (and maybe never will?)

### 🚀 Superintelligence (ASI)
- Surpasses human intelligence in all areas
- Science fiction (for now!)

```python
# Visualize the AI spectrum
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(12, 6))

# AI types on a timeline
ai_types = ['Rule-based\nsystems', 'Machine Learning', 'Deep Learning', 
            'Generative AI', 'AGI?', 'ASI?']
years = [1960, 1980, 2010, 2020, 2040, 2060]
capabilities = [10, 20, 50, 70, 90, 100]

# Plot
ax.plot(years, capabilities, 'b-', linewidth=2)
ax.scatter(years[:4], capabilities[:4], s=100, c='green', label='Exists')
ax.scatter(years[4:], capabilities[4:], s=100, c='red', alpha=0.5, label='Hypothetical')

# Annotation
for i, (year, cap, ai_type) in enumerate(zip(years, capabilities, ai_types)):
    ax.annotate(ai_type, (year, cap), textcoords="offset points", 
                xytext=(0,10), ha='center', fontsize=10)

ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Relative Capability (%)', fontsize=12)
ax.set_title('AI Development: From Rule-based to AGI?', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend()

# Mark "we are here"
current_year = 2024
ax.axvline(x=current_year, color='orange', linestyle='--', alpha=0.7)
ax.text(current_year, 85, 'We are here!', rotation=90, va='bottom', color='orange')

plt.tight_layout()
plt.show()
```

## The Turing Test: Can machines think?

Alan Turing (1950) proposed a test:
- A person chats with both a human and a machine
- If the person cannot tell them apart → The machine is "intelligent"

### Relevance for healthcare?

```python
# Simulate a simple Turing test
import random

def medical_turing_test():
    """Simple demonstration of the Turing test concept"""
    
    # Predefined responses
    ai_responses = {
        "headache": "Based on the symptoms, this could be tension headache. "
                   "Common triggers include stress, dehydration, and poor sleep. "
                   "If symptoms persist or worsen, contact a doctor.",
        "fever": "Fever is the body's natural response to infection. "
                "Drink plenty of fluids, rest, and use fever reducers as needed. "
                "Seek medical care if fever > 40°C or lasts > 3 days.",
        "default": "I understand your concern. Based on the description, I recommend "
                  "monitoring the symptoms and contacting your primary care physician if they worsen."
    }
    
    human_responses = {
        "headache": "Oh, headaches can be so annoying! I remember when I had migraines... "
                   "Have you tried lying in a dark room? That often helps me. "
                   "But if it doesn't go away, you should probably talk to your doctor.",
        "fever": "Ugh, fever is never fun. Last time I was sick I drank lots of tea with honey. "
                "Grandma always used to say you have to sweat it out! "
                "But if it sounds high, maybe you should call the emergency room?",
        "default": "Hmm, that doesn't sound good. I'm not a doctor, but "
                  "I think you should get it checked if you're worried. "
                  "Better to be safe than sorry!"
    }
    
    print("🤖 TURING TEST: Who is responding - AI or human?")
    print("=" * 50)
    
    symptom = input("Describe a symptom (e.g. 'headache' or 'fever'): ").lower()
    
    # Randomly choose whether AI or human responds
    is_ai = random.choice([True, False])
    
    if is_ai:
        response = ai_responses.get(symptom, ai_responses["default"])
    else:
        response = human_responses.get(symptom, human_responses["default"])
    
    print(f"\n💬 Response:\n{response}")
    
    guess = input("\n🤔 Do you think this is AI or human? ").lower()
    
    if (guess == "ai" and is_ai) or (guess == "human" and not is_ai):
        print("✅ Correct!")
    else:
        print(f"❌ Wrong! It was {'AI' if is_ai else 'human'}.")
    
    print("\n💡 Notice:")
    print("- AI: More structured, fact-based, professional")
    print("- Human: Personal anecdotes, emotions, less formal")
```

```python
# Run the test
medical_turing_test()
```

## 🧪 Hands-on: Your First AI

Let's create a simple rule-based "AI" for symptom assessment!

```python
class SimpleSymptomAI:
    """A very simple rule-based AI for symptom assessment"""
    
    def __init__(self):
        # Define simple rules
        self.symptom_rules = {
            'fever': {
                'high': 'Contact doctor immediately',
                'moderate': 'Rest and drink fluids, contact doctor if lasts > 3 days',
                'low': 'Monitor, use fever reducers as needed'
            },
            'headache': {
                'sudden_severe': 'Call 911 - may be serious',
                'chronic': 'Schedule doctor appointment',
                'mild': 'Rest, drink water, consider pain relievers'
            },
            'chest_pain': {
                'with_shortness_of_breath': 'Call 911 immediately',
                'with_exertion': 'Contact doctor same day',
                'muscular': 'Monitor, consider pain relievers'
            }
        }
    
    def assess(self, symptom, severity):
        """Provide simple assessment based on symptom and severity"""
        symptom = symptom.lower()
        
        if symptom in self.symptom_rules:
            if severity in self.symptom_rules[symptom]:
                return self.symptom_rules[symptom][severity]
            else:
                return "Contact primary care physician for assessment"
        else:
            return "Symptom not recognized - contact healthcare personnel"
    
    def interactive_consultation(self):
        """Run an interactive consultation"""
        print("🤖 Simple Symptom AI")
        print("=" * 40)
        print("Available symptoms: fever, headache, chest_pain")
        
        symptom = input("\nWhat is bothering you? ").lower()
        
        if symptom == 'fever':
            print("Temperature: 1) Under 38.5  2) 38.5-39.5  3) Over 39.5")
            choice = input("Choose (1/2/3): ")
            severity = ['low', 'moderate', 'high'][int(choice)-1]
        elif symptom == 'headache':
            print("Type: 1) Mild  2) Chronic  3) Sudden and severe")
            choice = input("Choose (1/2/3): ")
            severity = ['mild', 'chronic', 'sudden_severe'][int(choice)-1]
        elif symptom == 'chest_pain':
            print("Character: 1) Muscular  2) With exertion  3) With shortness of breath")
            choice = input("Choose (1/2/3): ")
            severity = ['muscular', 'with_exertion', 'with_shortness_of_breath'][int(choice)-1]
        else:
            severity = 'unknown'
        
        recommendation = self.assess(symptom, severity)
        
        print("\n" + "="*40)
        print(f"📋 Assessment: {recommendation}")
        print("="*40)
        print("\n⚠️ IMPORTANT: This is only a demonstration!")
        print("For real symptoms - always contact healthcare personnel!")
```

```python
# Create and test the AI
my_ai = SimpleSymptomAI()
```

```python
# Test with examples
print("📝 Test of rule-based AI:")
print(f"Fever (high): {my_ai.assess('fever', 'high')}")
print(f"Headache (mild): {my_ai.assess('headache', 'mild')}")
print(f"Chest pain (with_shortness_of_breath): {my_ai.assess('chest_pain', 'with_shortness_of_breath')}")
```
