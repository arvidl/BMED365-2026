## Why networks in medicine?

As a physician, you will encounter complex relationships daily: How are symptoms connected to diagnoses? Which patients respond similarly to a treatment? How do diseases spread in a population?

Traditionally, medicine has focused on **single variables** – one risk factor, one symptom, one treatment. But the body and healthcare system are **systems** where everything is interconnected. To understand these relationships, we need tools that can handle **relations** between many elements simultaneously.

This is where **graph theory** and **network science** come in.

## Graph theory – the mathematics behind relationships

**Graph theory** is a branch of mathematics that studies **graphs** – structures that describe how objects are connected to each other.

🔑 **Key idea:** A graph consists of:
- **Nodes** (vertices) – the objects we study
- **Edges** (lines) – the connections between them

| Medical example | Nodes | Edges |
|-----------------|-------|-------|
| Disease network | Diseases | Shared risk factors |
| Symptom network | Symptoms | Co-occurrence |
| Protein interactions | Proteins | Physical binding |
| Disease transmission | People | Transmission contact |

Graph theory gives us a **precise language** to describe and analyze such relationships mathematically.

## Network science – from theory to insight

**Network science** applies graph theory to real systems to discover patterns and structure.

Key questions network science can answer:

| Question | Network concept | Medical application |
|----------|-----------------|---------------------|
| *Who is most central?* | Centrality | Which genes are most important in a disease process? |
| *Are there groups?* | Clusters/communities | Are there patient subgroups with different prognoses? |
| *How robust is the system?* | Network robustness | What happens if a protein stops functioning? |
| *How do things spread?* | Diffusion | How do infections spread in a population? |

> 📊 **Fun fact:** Network science has revealed that many biological networks have a similar structure to social networks and the internet – so-called *scale-free* networks.

<details>
<summary>🔬 <b>What is a "scale-free" network?</b> (click to expand)</summary>

### Scale-free networks explained

In a **scale-free** network, most nodes have few connections, while a few nodes – called **hubs** – have very many connections.

**Intuitive example: Airports**
- Most airports have few direct routes
- But a few (London Heathrow, Frankfurt, Dubai) are *hubs* with hundreds of routes
- If a small airport closes, it's barely noticed
- If a hub closes, large parts of the network collapse

**In medicine and biology:**

| Network | Hubs (central nodes) | Consequence |
|---------|----------------------|-------------|
| Protein interactions | A few proteins bind to hundreds of others | Mutation in a hub → serious disease |
| Gene regulation | Master regulators (e.g., p53) | p53 mutation found in ~50% of cancer cases |
| Disease network | Some diseases share genes with many others | Explains comorbidity |
| Brain network | Hub regions (e.g., prefrontal cortex) | Damage here causes greater functional loss |

**Mathematically:** In a scale-free network, the degree distribution follows a *power law*: $P(k) \propto k^{-\gamma}$, where $k$ is the number of connections and $\gamma$ is typically between 2 and 3.

**Why is this important?**
1. **Robustness:** Scale-free networks tolerate random failures well (most nodes have few connections)
2. **Vulnerability:** But they are highly vulnerable to targeted attacks on hubs
3. **Drug targets:** Hubs are often good drug targets – but also risky (side effects)

> 💡 **Clinical insight:** When you see that a protein is involved in many diseases, it may be a hub in the biological network – and thus a potential drug target, but also a source of side effects.

## Patient Similarity Networks (PSN)

**Patient similarity networks** are a specific application of network science in clinical medicine.

### The idea is simple:
1. **Each patient is a node**
2. **Similar patients are connected by an edge**
3. **The more similar, the stronger the connection** (weighted edge)

### What are they used for?

| Application | Description |
|-------------|-------------|
| **Disease subtyping** | Finding patient subgroups with different disease mechanisms |
| **Prognosis** | Predicting outcomes based on similar patients |
| **Treatment selection** | "Patients like you responded best to treatment X" |
| **Rare diseases** | Finding similar patients across hospitals |

### Example: Cancer treatment
Imagine 1000 cancer patients. Instead of treating everyone the same, we can:
1. Calculate similarity based on genetics, biomarkers, and clinical data
2. Build a patient similarity network
3. Identify clusters of similar patients
4. See which treatment worked best in each cluster
5. Give personalized recommendations to new patients

> 🎯 **Precision medicine:** PSN is an important tool for moving from "one-size-fits-all" to personalized medicine.

## 📜 Historical perspective: From Königsberg to network medicine

Graph theory has a fascinating history that starts with an everyday problem:

### Euler's bridges (1736)
The mathematician **Leonhard Euler** studied a seemingly simple question: *Can one walk across all seven bridges of Königsberg exactly once?* By abstracting the problem to nodes (land areas) and edges (bridges), he proved it was impossible – and thereby founded graph theory.

> 🌉 **The lesson:** By focusing on *relationships* instead of details, Euler could solve a problem that had confused people for generations.

**The four land areas:**

- **B** – Kneiphof (the large island in the middle)
- **C** – Lomse (the small island/area to the right)  
- **D** – South bank (bottom)

**The question:** Can you start somewhere, cross *each* bridge *exactly once*, and end up somewhere (not necessarily the starting point)?

The inhabitants of Königsberg tried for years without success. Was it impossible, or had they just not found the route?

---

### Euler's abstraction: From map to graph

Euler's stroke of genius was to realize that the physical details (bridge lengths, island shapes, distances) were *irrelevant*. The only thing that mattered was the **topology** – who is connected to whom:

- **Nodes** $V = \{A, B, C, D\}$: The four land areas
- **Edges** $E$: The seven bridges

**The multigraph** (graph with multiple edges between the same nodes):

| Connection | Number of bridges |
|------------|-------------------|
| A — B | 2 |
| A — C | 1 |
| A — D | 1 |
| B — C | 1 |
| B — D | 1 |
| C — D | 1 |

**Total:** 7 edges (bridges)

---

### Formal definition: Eulerian path and Eulerian circuit

**Definition 1 (Eulerian path):**
An *Eulerian path* in a graph $G = (V, E)$ is a path that visits **each edge exactly once**.

**Definition 2 (Eulerian circuit):**
An *Eulerian circuit* is an Eulerian path that starts and ends at the same node (a closed path).

**Definition 3 (Degree):**
The *degree* of a node $v$, denoted $\deg(v)$, is the number of edges incident to $v$.

---

### Euler's theorem: Necessary and sufficient conditions

**Theorem (Euler, 1736):**
Let $G$ be a connected graph. Then:

1. $G$ has an **Eulerian circuit** $\Longleftrightarrow$ all nodes have **even degree**

2. $G$ has an **Eulerian path** (but not circuit) $\Longleftrightarrow$ exactly **two nodes** have **odd degree**

---

### Proof: Why Königsberg has no Eulerian path

**Step 1: Calculate the degree of each node**

$$\deg(A) = 2 + 1 + 1 = 3 \quad \text{(bridges to B, C, D)}$$
$$\deg(B) = 2 + 1 + 1 = 5 \quad \text{(bridges to A, C, D, but 2 to A)}$$

Wait – let's count exactly from the table:
- Node A is endpoint for: 2 (to B) + 1 (to C) + 1 (to D) = **3 edges**
- Node B is endpoint for: 2 (to A) + 1 (to C) + 1 (to D) = **5 edges**

No, this isn't right. Let me count again from the historical problem:

The seven bridges connect:
1. A — B (bridge 1)
2. A — B (bridge 2)  
3. A — C (bridge 3)
4. B — C (bridge 4)
5. B — D (bridge 5)
6. C — D (bridge 6)
7. C — D (bridge 7)

**Correct degree counting:**

| Node | Connected bridges | Degree |
|------|-------------------|--------|
| A | 1, 2, 3 | $\deg(A) = 3$ |
| B | 1, 2, 4, 5 | $\deg(B) = 5$ |
| C | 3, 4, 6, 7 | $\deg(C) = 3$ |
| D | 5, 6, 7 | $\deg(D) = 3$ |

**Verification:** $\sum_{v \in V} \deg(v) = 3 + 5 + 3 + 3 = 14 = 2 \cdot |E| = 2 \cdot 7$ ✓

(The sum of all degrees is always $2 \times$ the number of edges, since each edge contributes to the degree of exactly two nodes.)

---

**Step 2: Count nodes with odd degree**

| Node | Degree | Parity |
|------|--------|--------|
| A | 3 | Odd |
| B | 5 | Odd |
| C | 3 | Odd |
| D | 3 | Odd |

**Number of nodes with odd degree:** 4

---

**Step 3: Apply Euler's theorem**

For an Eulerian path to exist, we need:
- 0 nodes with odd degree (Eulerian circuit), or
- 2 nodes with odd degree (Eulerian path)

**But Königsberg has 4 nodes with odd degree!**

$$|\{v \in V : \deg(v) \text{ is odd}\}| = 4$$

$$4 \neq 0 \quad \text{and} \quad 4 \neq 2$$

---

**Step 4: Conclusion**

$$\boxed{\text{The Königsberg graph has no Eulerian path}}$$

It is **mathematically impossible** to cross all seven bridges exactly once. $\blacksquare$

---

### The intuition behind the proof

**Why must odd-degree nodes be ≤ 2?**

Think about what happens when you walk along a path:

1. **Start node:** You *leave* without having arrived → use 1 edge (odd contribution)
2. **Intermediate nodes:** You arrive and leave → use 2 edges per visit (even contribution)
3. **End node:** You *arrive* without leaving → use 1 edge (odd contribution)

If start ≠ end: exactly 2 nodes have "unbalanced" edge usage (odd degree)
If start = end: all nodes have "balanced" edge usage (even degree)

**In Königsberg** ALL four nodes have odd degree. This means that *four* nodes would need to be either start or end – but a path can only have one start and one end!

### Medical implications

Euler's approach – abstracting a problem to nodes and edges – has direct applications in medicine:

#### 1. **Metabolic pathways**
Metabolites (nodes) connected by enzymatic reactions (edges). Questions like "can we get from substrate A to product B?" are solved with graph theory.

#### 2. **Blood vessels and nerve pathways**
- Can blood/nerve impulses reach all areas?
- If a node (blood vessel/nerve) is blocked, are there alternative routes?
- **Eulerian path thinking:** Can a surgeon plan a route that visits all structures without crossing the same place twice?

#### 3. **DNA sequencing**
De Bruijn graphs (a type of Eulerian graph) are used in modern DNA sequencing to reconstruct genomes from short fragments.

#### 4. **Epidemiology: Transmission routes**
- Nodes = people/places
- Edges = transmission contact
- Question: "Can the infection reach everyone?" corresponds to graph connectivity
- "How many contacts must be broken to stop spread?" corresponds to edge removal

#### 5. **Drug interactions**
- Nodes = drugs
- Edges = known interactions
- "Is there a 'path' of interactions from drug A to side effect X?"

> 🧠 **Paradigm shift:** Euler showed that some problems that *look* geographic/physical are actually *topological* – they are about connections, not distances. This perspective shift is the core of network medicine.

#### Example: Brain connectivity

In brain research, scientists study whether the brain has an "Euler-like" structure:
- **Nodes:** Brain regions
- **Edges:** Nerve connections (white matter)
- **Question:** Is the brain efficiently "wired"? 

Studies show that the brain has a *small-world* structure – it is neither completely random nor completely regular, but optimized for efficient communication with minimal "wiring".

## ⚠️ Limitations and challenges

Network analysis is powerful, but not always the right tool:

| Challenge | Description |
|-----------|-------------|
| **Data quality** | "Garbage in, garbage out" – networks are only as good as the data |
| **Causality vs. correlation** | An edge means *association*, not necessarily *causation* |
| **Complexity** | Large networks can be difficult to interpret |
| **Privacy** | Patient networks contain sensitive information |
| **Over-interpretation** | Not everything is a network – sometimes simpler models are better |

> 🤔 **Critical thinking:** Always ask: *What do the nodes and edges actually represent?* and *Is this network representation meaningful for my problem?*

## What is a graph?

A **graph** is a mathematical structure that shows relationships between objects. It consists of:

| Term | Description | Examples |
|------|-------------|----------|
| **Nodes** (vertices) | The objects we study | Patients, diseases, symptoms |
| **Edges** | Connections between nodes | Similarity, relation, interaction |

````python
# ============================================================================
# A SIMPLE EXAMPLE: CREATE YOUR FIRST GRAPH
# ============================================================================

import networkx as nx
import matplotlib.pyplot as plt

# Create a graph object
graph = nx.Graph()

# Add nodes
graph.add_node("Patient A")
graph.add_node("Disease X")

# Add an edge (connection)
graph.add_edge("Patient A", "Disease X")

# Print the nodes and edges
print("Nodes:", list(graph.nodes()))
print("Edges:", list(graph.edges()))
````

````python
# Visualize the graph
plt.figure(figsize=(6, 4))
nx.draw(graph, 
        with_labels=True, 
        node_size=2000, 
        node_color="lightblue", 
        font_size=12, 
        font_weight="bold",
        edge_color="gray",
        width=2)
plt.title("A simple graph: Patient-Disease relationship")
plt.show()

print("✓ Your first graph is created!")
````

### The relationship between graph and adjacency matrix

An **adjacency matrix** is a compact way to represent a graph:

| Property | Explanation |
|----------|-------------|
| **Rows** | Represent the **from-node** (starting point of the edge) |
| **Columns** | Represent the **to-node** (endpoint of the edge) |
| **Value = 1** | There is an edge from the row-node to the column-node |
| **Value = 0 (empty)** | No edge from the row-node to the column-node |

**For the undirected graph:**
- The matrix is **symmetric** around the diagonal
- If there is an edge a→b, there is also b→a
- Example: `matrix[a,b] = matrix[b,a] = 1`

**For the directed graph:**
- The matrix is **not symmetric**
- The edge goes only in the arrow's direction: from row to column
- Example: If `matrix[a,b] = 1` but `matrix[b,a] = 0`, it means there is an edge *from* a *to* b, but *not* from b to a

> 💡 **Clinical relevance:** In patient similarity networks, we typically use undirected graphs because similarity between patients is symmetric – if patient A resembles patient B, B also resembles A.

## The adjacency matrix

An **adjacency matrix** is a way to represent a graph as a table:
- The rows and columns represent the nodes
- The value in cell (i,j) indicates whether there is an edge between node i and node j

This is important because many algorithms use matrix calculations!


````python
# ============================================================================
# ADJACENCY MATRICES AS TEXT
# ============================================================================

def print_adjacency_matrix(G, title):
    """Print the adjacency matrix for a graph."""
    nodes = sorted(G.nodes())
    print(f"\n{title}:")
    print("  " + " ".join(nodes))
    for i in nodes:
        row = [str(1) if G.has_edge(i, j) else " " for j in nodes]
        print(f"{i} {' '.join(row)}")

print_adjacency_matrix(G_undirected, "Undirected adjacency matrix")
print_adjacency_matrix(G_directed, "Directed adjacency matrix")
````

````python
# ============================================================================
# ADJACENCY MATRICES AS GRAPHICS
# ============================================================================

import numpy as np

# Create figure with 2x2 subplots
fig = plt.figure(figsize=(15, 10))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 1])

def create_adjacency_matrix(G):
    """Convert graph to numpy adjacency matrix."""
    nodes = sorted(G.nodes())
    n = len(nodes)
    matrix = np.zeros((n, n))
    for i, node1 in enumerate(nodes):
        for j, node2 in enumerate(nodes):
            if G.has_edge(node1, node2):
                matrix[i, j] = 1
    return matrix, nodes

def draw_adjacency_matrix(ax, matrix, nodes, title):
    """Draw adjacency matrix as grid."""
    n = len(nodes)
    ax.matshow(matrix, cmap='binary')
    
    # Add grid
    for i in range(n+1):
        ax.axhline(i-0.5, color='black', linewidth=1)
        ax.axvline(i-0.5, color='black', linewidth=1)
    
    # Add values
    for i in range(n):
        for j in range(n):
            if matrix[i, j] == 1:
                ax.text(j, i, '1', ha='center', va='center', fontsize=12)
    
    # Add node names
    ax.set_xticks(range(n))
    ax.set_yticks(range(n))
    ax.set_xticklabels(nodes)
    ax.set_yticklabels(nodes)
    ax.tick_params(axis='both', which='both', length=0)
    ax.set_title(title)

# Draw undirected graph
ax1 = fig.add_subplot(gs[0, 0])
nx.draw(G_undirected, pos, ax=ax1,
        with_labels=True,
        node_color='lightblue',
        node_size=500,
        font_size=12,
        font_weight='bold')
ax1.set_title("Undirected, binary graph")

# Draw directed graph
ax2 = fig.add_subplot(gs[0, 1])
nx.draw(G_directed, pos, ax=ax2,
        with_labels=True,
        node_color='lightblue',
        node_size=500,
        font_size=12,
        font_weight='bold',
        arrows=True,
        arrowsize=20)
ax2.set_title("Directed, binary graph")

# Draw undirected adjacency matrix
ax3 = fig.add_subplot(gs[1, 0])
matrix_undirected, nodes = create_adjacency_matrix(G_undirected)
draw_adjacency_matrix(ax3, matrix_undirected, nodes, "Undirected adjacency matrix")

# Draw directed adjacency matrix
ax4 = fig.add_subplot(gs[1, 1])
matrix_directed, nodes = create_adjacency_matrix(G_directed)
draw_adjacency_matrix(ax4, matrix_directed, nodes, "Directed adjacency matrix")

plt.tight_layout()
plt.show()

print("✓ Graphs and adjacency matrices are visualized!")
````

### Weighted graphs and adjacency matrices

In many medical applications, not all connections are equally strong. For example:
- **Patient similarity networks**: The similarity between patients can vary from 0 (no similarity) to 1 (identical)
- **Protein interactions**: The binding strength varies between proteins
- **Symptom correlation**: Some symptoms co-occur more often than others

In a **weighted graph**, each edge has a numerical value (weight) that represents the strength of the connection. In the adjacency matrix, this is shown as values between 0 and 1 (or other scales), where higher values indicate stronger connections.

> 💡 **Note:** Edges with weight = 0 are not shown in the graph (no connection).


````python
# ============================================================================
# WEIGHTED GRAPHS AND ADJACENCY MATRICES
# ============================================================================

import numpy as np
import warnings

# --- Define weighted graphs with the same structure as before ---

# Undirected weighted graph (same edges as binary version, but with weights)
G_undirected_weighted = nx.Graph()
G_undirected_weighted.add_weighted_edges_from([
    ('a', 'b', 0.9),   # Strong connection
    ('b', 'c', 0.5),   # Medium connection
    ('c', 'd', 0.3),   # Weaker connection
    ('c', 'e', 0.8),   # Strong connection
    ('d', 'e', 0.6)    # Medium connection
])

# Directed weighted graph (same edges as binary version, but with weights)
G_directed_weighted = nx.DiGraph()
G_directed_weighted.add_weighted_edges_from([
    ('a', 'b', 0.7),   # a → b
    ('b', 'a', 0.4),   # b → a (asymmetric!)
    ('c', 'b', 0.9),   # c → b
    ('c', 'd', 0.2),   # c → d (weak)
    ('c', 'e', 0.8),   # c → e
    ('e', 'd', 0.5),   # e → d
    ('e', 'c', 0.3)    # e → c (weaker back)
])

# Use same positions as before
pos = {
    'a': (-1, 0),
    'b': (-0.5, 1),
    'c': (0, 0),
    'd': (1, 1),
    'e': (1, 0)
}

# ============================================================================
# FUNCTIONS FOR WEIGHTED ADJACENCY MATRICES
# ============================================================================

def create_weighted_adjacency_matrix(G):
    """Convert weighted graph to numpy adjacency matrix with weights."""
    nodes = sorted(G.nodes())
    n = len(nodes)
    matrix = np.zeros((n, n))
    for i, node1 in enumerate(nodes):
        for j, node2 in enumerate(nodes):
            if G.has_edge(node1, node2):
                # Get the weight from the edge
                matrix[i, j] = G[node1][node2].get('weight', 1.0)
    return matrix, nodes

def draw_weighted_adjacency_matrix(ax, matrix, nodes, title):
    """Draw weighted adjacency matrix with grayscale."""
    n = len(nodes)
    
    # Use grayscale (0=white, 1=black)
    ax.matshow(matrix, cmap='Greys', vmin=0, vmax=1)
    
    # Add grid
    for i in range(n+1):
        ax.axhline(i-0.5, color='gray', linewidth=0.5)
        ax.axvline(i-0.5, color='gray', linewidth=0.5)
    
    # Add values (weights)
    for i in range(n):
        for j in range(n):
            if matrix[i, j] > 0:
                # Choose text color based on background
                text_color = 'white' if matrix[i, j] > 0.5 else 'black'
                ax.text(j, i, f'{matrix[i, j]:.1f}', 
                       ha='center', va='center', fontsize=10, 
                       color=text_color, fontweight='bold')
    
    # Add node names
    ax.set_xticks(range(n))
    ax.set_yticks(range(n))
    ax.set_xticklabels(nodes)
    ax.set_yticklabels(nodes)
    ax.tick_params(axis='both', which='both', length=0)
    ax.set_title(title)

# ============================================================================
# VISUALIZATION
# ============================================================================

fig = plt.figure(figsize=(15, 10))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 1], hspace=0.3)

# --- Draw undirected weighted graph ---
ax1 = fig.add_subplot(gs[0, 0])
# Get weights for edge thickness
weights_undirected = [G_undirected_weighted[u][v]['weight'] * 4 for u, v in G_undirected_weighted.edges()]
nx.draw(G_undirected_weighted, pos, ax=ax1,
        with_labels=True,
        node_color='lightgreen',
        node_size=700,
        font_size=12,
        font_weight='bold',
        width=weights_undirected,  # Edge thickness based on weight
        edge_color='darkgreen')
# Add edge weights as labels
edge_labels_undirected = {(u, v): f'{d["weight"]:.1f}' for u, v, d in G_undirected_weighted.edges(data=True)}
nx.draw_networkx_edge_labels(G_undirected_weighted, pos, edge_labels_undirected, ax=ax1, font_size=9)
ax1.set_title("Undirected, weighted graph", fontsize=12)

# --- Draw directed weighted graph ---
ax2 = fig.add_subplot(gs[0, 1])
# Get weights for edge thickness
weights_directed = [G_directed_weighted[u][v]['weight'] * 4 for u, v in G_directed_weighted.edges()]
nx.draw(G_directed_weighted, pos, ax=ax2,
        with_labels=True,
        node_color='lightyellow',
        node_size=700,
        font_size=12,
        font_weight='bold',
        width=weights_directed,  # Edge thickness based on weight
        edge_color='darkorange',
        arrows=True,
        arrowsize=15,
        connectionstyle='arc3,rad=0.1')  # Curve to show both directions
# Add edge weights as labels
edge_labels_directed = {(u, v): f'{d["weight"]:.1f}' for u, v, d in G_directed_weighted.edges(data=True)}
nx.draw_networkx_edge_labels(G_directed_weighted, pos, edge_labels_directed, ax=ax2, font_size=9)
ax2.set_title("Directed, weighted graph", fontsize=12)

# --- Draw undirected weighted adjacency matrix ---
ax3 = fig.add_subplot(gs[1, 0])
matrix_undirected_w, nodes = create_weighted_adjacency_matrix(G_undirected_weighted)
draw_weighted_adjacency_matrix(ax3, matrix_undirected_w, nodes, "Undirected, weighted adjacency matrix")

# --- Draw directed weighted adjacency matrix ---
ax4 = fig.add_subplot(gs[1, 1])
matrix_directed_w, nodes = create_weighted_adjacency_matrix(G_directed_weighted)
draw_weighted_adjacency_matrix(ax4, matrix_directed_w, nodes, "Directed, weighted adjacency matrix")

# Add color scale
cbar_ax = fig.add_axes([0.92, 0.15, 0.02, 0.3])
sm = plt.cm.ScalarMappable(cmap='Greys', norm=plt.Normalize(vmin=0, vmax=1))
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax)
cbar.set_label('Weight (connection strength)', fontsize=10)

# Suppress warning about tight_layout and colorbar axis
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    plt.tight_layout(rect=[0, 0, 0.9, 1])
plt.show()

print("✓ Weighted graphs and adjacency matrices visualized!")
print()
print("Note:")
print("  • Thicker edges = stronger connections")
print("  • Darker cells in the adjacency matrix = higher weight")
print("  • In the directed graph: a→b (0.7) ≠ b→a (0.4) – asymmetric!")
````

In the graphs, this is shown as edge thickness and numerical labels. In the adjacency matrices, grayscale is used where darker color = stronger connection.*

## 📍 Centrality measures: Who is most important in the network?

In a network, not all nodes are equally important. Think of a social network: some people have many friends, others function as "bridges" between different groups, and some know "all the right people." **Centrality measures** are mathematical tools that help us identify the most important nodes in a network.

### 🏥 Why is this important in medicine?

| Application | Questions centrality measures can answer |
|-------------|------------------------------------------|
| **Epidemiology** | Who are the super-spreaders that should be prioritized for vaccination? |
| **Drug development** | Which proteins are most critical to target? |
| **Patient similarity** | Which patients are "typical" for a disease group? |
| **Disease network** | Which diseases connect many other conditions? |

### 🔢 The four most important centrality measures

<p><strong>Question:</strong> How many direct connections does the node have?</p>

#### 📐 Mathematical definitions

| Centrality measure | Formula | Explanation |
|--------------------|---------|-------------|
| **Degree centrality** | $$C_D(v) = \frac{\deg(v)}{n-1}$$ | Number of neighbors normalized |
| **Betweenness centrality** | $$C_B(v) = \sum_{s \neq v \neq t} \frac{\sigma_{st}(v)}{\sigma_{st}}$$ | Proportion of shortest paths through $v$ |
| **Closeness centrality** | $$C_C(v) = \frac{n-1}{\sum_{u \neq v} d(v,u)}$$ | Inverse average distance |
| **Eigenvector centrality** | $$\mathbf{A}\mathbf{x} = \lambda \mathbf{x}$$ | Solution of eigenvalue problem |

where:
- $\deg(v)$ = number of edges to node $v$
- $n$ = total number of nodes
- $\sigma_{st}$ = number of shortest paths from $s$ to $t$
- $\sigma_{st}(v)$ = number of these that pass through $v$
- $d(v,u)$ = shortest distance from $v$ to $u$
- $\mathbf{A}$ = adjacency matrix
- $\mathbf{x}$ = eigenvector (centrality values for all nodes)
- $\lambda$ = largest eigenvalue

> 💡 **About eigenvector centrality:** The formula $x_v = \frac{1}{\lambda} \sum_{u \in N(v)} x_u$ *looks* recursive, but is solved as an eigenvalue problem: we find the eigenvector $\mathbf{x}$ of the adjacency matrix $\mathbf{A}$ that belongs to the largest eigenvalue $\lambda$. Intuitively: a node's centrality is proportional to the sum of its neighbors' centrality – you are important if you know important people.

### 🎯 An intuitive example

Think of a hospital as a network where nodes are departments and edges are patient transfers:

```
                    ┌──────────────┐
                    │   Intensive  │ ← High betweenness centrality
                    │  Care Unit   │   (critical bridge)
                    └──────┬───────┘
                           │
          ┌────────────────┼────────────────┐
          │                │                │
    ┌─────┴─────┐    ┌─────┴─────┐    ┌─────┴─────┐
    │  Surgery  │    │  Medicine │    │ Emergency │
    │           │    │           │    │   Room    │ ← High degree centrality
    └─────┬─────┘    └───────────┘    └─────┬─────┘   (many connections)
          │                                  │
    ┌─────┴─────┐                      ┌─────┴─────┐
    │ Orthopedi │                      │  Triage   │
    └───────────┘                      └───────────┘
```

- **Emergency Room** has high *degree centrality* (many patients come in/out)
- **Intensive Care Unit** has high *betweenness centrality* (connects different specialties)
- A department with high *closeness centrality* can quickly reach all others
- A department with high *eigenvector centrality* is connected to other important departments

### 💡 Practical significance

By identifying central nodes, we can:
- **Optimize resources**: Focus on critical points
- **Predict vulnerabilities**: What happens if a central node fails?
- **Target interventions**: Vaccinate super-spreaders first
- **Understand the system**: Reveal hidden structures in complex data
